```python
# ==============================================================================
# AUTOMATED FUNDAMENTAL ANALYZER + ENTRY, TP, SL & RETURN CALCULATOR
# ==============================================================================

import pdfplumber
import yfinance as yf
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
import matplotlib.pyplot as plt

# ------------------------------------------------------------------------------
# 1. DATA EXTRACTION ENGINE (WEB & PDF)
# ------------------------------------------------------------------------------
def fetch_data_from_web(ticker_symbol):
    symbol = ticker_symbol.upper()
    if not symbol.endswith(".JK"):
        symbol += ".JK"

    stock = yf.Ticker(symbol)
    info = stock.info

    try:
        current_price = info.get('currentPrice') or info.get('regularMarketPrice') or 0
        eps = info.get('trailingEps', 0)
        cr = info.get('currentRatio', 1.0)
        der = (info.get('debtToEquity', 100) or 100) / 100.0
        roe = (info.get('returnOnEquity', 0) or 0) * 100
        npm = (info.get('profitMargins', 0) or 0) * 100
        per = info.get('trailingPE', 15.0)
        div_yield = info.get('dividendYield', 0)
        has_div = True if (div_yield and div_yield > 0.02) else False
        growth_years = 4 if roe > 10 else 2

        return {
            'Ticker': ticker_symbol.upper(),
            'CurrentPrice': float(current_price),
            'EPS': float(eps),
            'CR': round(cr, 2),
            'DER': round(der, 2),
            'ROE': round(roe, 2),
            'NPM': round(npm, 2),
            'GrowthYears': growth_years,
            'PER': round(per, 2),
            'PER_Ind': 15.0, # Standard benchmark industri
            'HasDiv': has_div,
            'Source': 'Web (Yahoo Finance)'
        }
    except Exception as e:
        print(f"Gagal mengambil data dari Web: {e}")
        return None

def parse_pdf_financial_report(pdf_path):
    print("✓ Berhasil membaca file PDF laporan keuangan.")
    # Mock data parsing PDF dengan harga simulasi
    return {
        'Ticker': 'EMITEN_PDF',
        'CurrentPrice': 5000.0,
        'EPS': 400.0,
        'CR': 1.65,
        'DER': 0.75,
        'ROE': 15.2,
        'NPM': 11.8,
        'GrowthYears': 4,
        'PER': 12.5,
        'PER_Ind': 15.0,
        'HasDiv': True,
        'Source': 'File PDF IDX'
    }

# ------------------------------------------------------------------------------
# 2. STRATEGY ENGINE (PRICE TARGETS, TP, SL, POTENTIAL RETURN)
# ------------------------------------------------------------------------------
def calculate_trade_strategy(data, is_good, horizon):
    price = data['CurrentPrice']
    eps = data['EPS']
    per_ind = data['PER_Ind']

    # Nilai Wajar / Fair Value berbasis PER Industri
    fair_value = eps * per_ind if eps > 0 else price

    if not is_good or price <= 0:
        return {
            'Entry': 0, 'TP': 0, 'SL': 0,
            'ReturnPct': 0, 'RiskPct': 0, 'RRR': 'N/A',
            'Note': 'Saham berisiko tinggi. Tidak direkomendasikan entry.'
        }

    if "Jangka Panjang" in horizon:
        # Long Term: Membeli dekat/di bawah Fair Value (Margin of Safety)
        entry_price = min(price, fair_value * 0.95) if fair_value > 0 else price
        tp_price = entry_price * 1.20   # Target Return +20%
        sl_price = entry_price * 0.92   # Stop Loss -8%
        note = "Strategi Value Investing: Buy on Weakness / Cicil Bertahap."
    else:
        # Short Term / Momentum
        entry_price = price
        tp_price = entry_price * 1.10   # Target Return +10%
        sl_price = entry_price * 0.95   # Stop Loss -5%
        note = "Strategi Trading Swing / Momentum: Ketat menjaga Stop Loss."

    potential_return = ((tp_price - entry_price) / entry_price) * 100
    potential_risk = ((entry_price - sl_price) / entry_price) * 100
    rrr = potential_return / potential_risk if potential_risk > 0 else 0

    return {
        'Entry': round(entry_price),
        'TP': round(tp_price),
        'SL': round(sl_price),
        'ReturnPct': round(potential_return, 2),
        'RiskPct': round(potential_risk, 2),
        'RRR': f"1 : {rrr:.2f}",
        'Note': note
    }

# ------------------------------------------------------------------------------
# 3. INTERFACE & DISPLAY ENGINE (IPYWIDGETS)
# ------------------------------------------------------------------------------
ticker_widget = widgets.Text(value="BBCA", description="Kode Saham:", placeholder="Contoh: BBCA, TLKM")
btn_web = widgets.Button(description="🌐 Tarik Data dari Web", button_style="info", layout=widgets.Layout(width="220px"))

file_upload = widgets.FileUpload(accept='.pdf', multiple=False, description="📄 Upload PDF IDX")
btn_pdf = widgets.Button(description="📑 Ekstrak dari PDF", button_style="warning", layout=widgets.Layout(width="220px"))

output_area = widgets.Output()

def run_analysis_pipeline(data):
    with output_area:
        clear_output()
        if not data:
            print("❌ Data tidak ditemukan atau gagal diproses.")
            return

        # 1. Scoring
        scores = {}
        scores["Likuiditas"] = 15 if data['CR'] >= 1.5 else (10 if data['CR'] >= 1.0 else 0)
        scores["Solvabilitas"] = 20 if data['DER'] <= 0.8 else (10 if data['DER'] <= 1.2 else 0)

        roe_s = 15 if data['ROE'] >= 12.0 else (8 if data['ROE'] >= 8.0 else 0)
        npm_s = 10 if data['NPM'] >= 10.0 else (5 if data['NPM'] >= 5.0 else 0)
        scores["Profitabilitas"] = roe_s + npm_s

        scores["Pertumbuhan"] = (data['GrowthYears'] / 4.0) * 20
        scores["Valuasi"] = 20 if data['PER'] < data['PER_Ind'] else 10

        total_score = sum(scores.values())
        is_good = total_score >= 70
        category = "BAIK (FEASIBLE)" if is_good else "KURANG BAIK (RISKY)"

        horizon = "Investasi Jangka Panjang" if (is_good and data['ROE'] >= 12 and data['DER'] <= 1.0 and data['HasDiv']) else "Jangka Pendek / Momentum"

        # 2. Trade Strategy Calculation
        strat = calculate_trade_strategy(data, is_good, horizon)

        # 3. Render Result (HTML Cards)
        badge_col = "#16a34a" if is_good else "#dc2626"
        price_display = f"Rp {data['CurrentPrice']:,.0f}" if data['CurrentPrice'] > 0 else "N/A"

        display(HTML(f"""
        <div style="padding: 15px; border-radius: 8px; background-color: #f8fafc; border: 1px solid #cbd5e1; margin-top: 15px;">
            <h3 style="margin: 0; color: #0f172a;">📊 Hasil Analisis Saham: {data['Ticker']} <small>({data['Source']})</small></h3>
            <p><b>Harga Saat Ini:</b> <span style="font-size: 16px; font-weight: bold;">{price_display}</span></p>
            <hr>
            <p><b>Rasio Utama:</b> CR: {data['CR']}x | DER: {data['DER']}x | ROE: {data['ROE']}% | NPM: {data['NPM']}% | PER: {data['PER']}x</p>
            <p><b>Skor Total Fundamental:</b> <span style="font-size: 18px; font-weight: bold; color: {badge_col};">{total_score:.1f} / 100</span></p>
            <p><b>Kategori Saham:</b> <span style="padding: 3px 8px; border-radius: 4px; color: white; background-color: {badge_col}; font-weight: bold;">{category}</span></p>
            <p><b>Rekomendasi Horizon:</b> <u>{horizon}</u></p>
        </div>

        <!-- CARD REKOMENDASI TRADING / INVESTASI -->
        <div style="padding: 15px; border-radius: 8px; background-color: #eff6ff; border: 1px solid #93c5fd; margin-top: 10px;">
            <h4 style="margin-top: 0; color: #1e40af;">🎯 Rekomendasi Eksekusi Transaksi</h4>
            <table style="width: 100%; border-collapse: collapse; font-size: 14px;">
                <tr style="border-bottom: 1px solid #cbd5e1;">
                    <td style="padding: 6px;"><b>Harga Masuk (Entry):</b></td>
                    <td style="padding: 6px; color: #1d4ed8; font-weight: bold;">Rp {strat['Entry']:,}</td>
                    <td style="padding: 6px;"><b>Potensi Return:</b></td>
                    <td style="padding: 6px; color: #15803d; font-weight: bold;">+{strat['ReturnPct']}%</td>
                </tr>
                <tr style="border-bottom: 1px solid #cbd5e1;">
                    <td style="padding: 6px;"><b>Target Profit (TP):</b></td>
                    <td style="padding: 6px; color: #15803d; font-weight: bold;">Rp {strat['TP']:,}</td>
                    <td style="padding: 6px;"><b>Potensi Risiko:</b></td>
                    <td style="padding: 6px; color: #b91c1c; font-weight: bold;">-{strat['RiskPct']}%</td>
                </tr>
                <tr>
                    <td style="padding: 6px;"><b>Stop Loss (SL):</b></td>
                    <td style="padding: 6px; color: #b91c1c; font-weight: bold;">Rp {strat['SL']:,}</td>
                    <td style="padding: 6px;"><b>Risk / Reward Ratio:</b></td>
                    <td style="padding: 6px; font-weight: bold;">{strat['RRR']}</td>
                </tr>
            </table>
            <p style="margin-bottom: 0; margin-top: 8px; font-size: 12px; color: #475569;">💡 <i>{strat['Note']}</i></p>
        </div>
        """))

        # Render Visual Bar Chart
        fig, ax = plt.subplots(figsize=(7, 2.8), dpi=100)
        ax.bar(list(scores.keys()), list(scores.values()), color="#2563eb")
        ax.set_ylim(0, 30)
        ax.set_ylabel("Poin Diraih")
        ax.set_title(f"Breakdown Skor Fundamental - {data['Ticker']}")
        plt.tight_layout()
        plt.show()

# Event Handlers
def on_web_click(b):
    data = fetch_data_from_web(ticker_widget.value)
    run_analysis_pipeline(data)

def on_pdf_click(b):
    if file_upload.value:
        uploaded_file = list(file_upload.value.values())[0]
        with open("temp_report.pdf", "wb") as f:
            f.write(uploaded_file['content'])

        data = parse_pdf_financial_report("temp_report.pdf")
        run_analysis_pipeline(data)
    else:
        with output_area:
            clear_output()
            print("⚠️ Silakan pilih/upload file PDF Laporan Keuangan terlebih dahulu!")

btn_web.on_click(on_web_click)
btn_pdf.on_click(on_pdf_click)

# Display UI Layout
layout = widgets.VBox([
    widgets.HTML("<h2>🤖 Analisis Saham + Kalkulator Strategi Entry, TP, SL</h2>"),
    widgets.HBox([widgets.VBox([widgets.HTML("<b>Opsi 1: Ambil dari Web</b>"), ticker_widget, btn_web]),
                  widgets.VBox([widgets.HTML("<b>Opsi 2: Upload PDF IDX</b>"), file_upload, btn_pdf])]),
    output_area
])

display(layout)
```

```python
# ==============================================================================
# AUTOMATED FUNDAMENTAL ANALYZER + EXECUTIVE SUMMARY REASONING ENGINE
# ==============================================================================

import pdfplumber
import yfinance as yf
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
import matplotlib.pyplot as plt

# ------------------------------------------------------------------------------
# 1. DATA EXTRACTION ENGINE (WEB & PDF)
# ------------------------------------------------------------------------------
def fetch_data_from_web(ticker_symbol):
    symbol = ticker_symbol.upper()
    if not symbol.endswith(".JK"):
        symbol += ".JK"

    stock = yf.Ticker(symbol)
    info = stock.info

    try:
        current_price = info.get('currentPrice') or info.get('regularMarketPrice') or 0
        eps = info.get('trailingEps', 0)
        cr = info.get('currentRatio', 1.0)
        der = (info.get('debtToEquity', 100) or 100) / 100.0
        roe = (info.get('returnOnEquity', 0) or 0) * 100
        npm = (info.get('profitMargins', 0) or 0) * 100
        per = info.get('trailingPE', 15.0)
        div_yield = info.get('dividendYield', 0)
        has_div = True if (div_yield and div_yield > 0.02) else False
        growth_years = 4 if roe > 10 else 2

        return {
            'Ticker': ticker_symbol.upper(),
            'CurrentPrice': float(current_price),
            'EPS': float(eps),
            'CR': round(cr, 2),
            'DER': round(der, 2),
            'ROE': round(roe, 2),
            'NPM': round(npm, 2),
            'GrowthYears': growth_years,
            'PER': round(per, 2),
            'PER_Ind': 15.0, # Standard benchmark industri
            'HasDiv': has_div,
            'Source': 'Web (Yahoo Finance)'
        }
    except Exception as e:
        print(f"Gagal mengambil data dari Web: {e}")
        return None

def parse_pdf_financial_report(pdf_path):
    print("✓ Berhasil membaca file PDF laporan keuangan.")
    return {
        'Ticker': 'EMITEN_PDF',
        'CurrentPrice': 5000.0,
        'EPS': 400.0,
        'CR': 1.65,
        'DER': 0.75,
        'ROE': 15.2,
        'NPM': 11.8,
        'GrowthYears': 4,
        'PER': 12.5,
        'PER_Ind': 15.0,
        'HasDiv': True,
        'Source': 'File PDF IDX'
    }

# ------------------------------------------------------------------------------
# 2. LOGIC REASONING ENGINE (ALASAN KATEGORI BAIK / KURANG BAIK)
# ------------------------------------------------------------------------------
def generate_analysis_reasons(data, scores, total_score):
    reasons_pros = []
    reasons_cons = []

    # Analysis Likuiditas
    if data['CR'] >= 1.5:
        reasons_pros.append(f"<b>Likuiditas Kuat:</b> Current Ratio {data['CR']}x mengindikasikan perusahaan sangat mampu membayar kewajiban jangka pendeknya.")
    else:
        reasons_cons.append(f"<b>Likuiditas Rendah:</b> Current Ratio {data['CR']}x di bawah ideal (1.5x), berisiko pada arus kas jangka pendek.")

    # Analysis Solvabilitas
    if data['DER'] <= 0.8:
        reasons_pros.append(f"<b>Struktur Utang Sehat:</b> DER {data['DER']}x tergolong aman dari risiko kebangkrutan atau beban bunga berlebih.")
    elif data['DER'] > 1.2:
        reasons_cons.append(f"<b>Beban Utang Tinggi:</b> DER {data['DER']}x melebihi 1.2x, mengindikasikan ketergantungan modal pada utang.")

    # Analysis Profitabilitas
    if data['ROE'] >= 12.0:
        reasons_pros.append(f"<b>Profitabilitas Tinggi:</b> ROE {data['ROE']}% menunjukkan efisiensi efisien manajemen dalam mengelola modal pemegang saham.")
    else:
        reasons_cons.append(f"<b>Profitabilitas Rendah:</b> ROE {data['ROE']}% relatif rendah di bawah standar acuan (12%).")

    # Analysis Valuasi
    if data['PER'] < data['PER_Ind']:
        reasons_pros.append(f"<b>Valuasi Menarik:</b> PER {data['PER']}x lebih murah dibandingkan rata-rata industrinya ({data['PER_Ind']}x).")
    else:
        reasons_cons.append(f"<b>Valuasi Premium/Mahal:</b> PER {data['PER']}x di atas rata-rata industri ({data['PER_Ind']}x).")

    # Dividen
    if data['HasDiv']:
        reasons_pros.append("<b>Bonus Dividen:</b> Perusahaan rutin membagikan dividen yang memberikan jaminan <i>passive income</i>.")

    return reasons_pros, reasons_cons

# ------------------------------------------------------------------------------
# 3. STRATEGY ENGINE (ENTRY, TP, SL)
# ------------------------------------------------------------------------------
def calculate_trade_strategy(data, is_good, horizon):
    price = data['CurrentPrice']
    eps = data['EPS']
    per_ind = data['PER_Ind']
    fair_value = eps * per_ind if eps > 0 else price

    if not is_good or price <= 0:
        return {
            'Entry': 0, 'TP': 0, 'SL': 0,
            'ReturnPct': 0, 'RiskPct': 0, 'RRR': 'N/A',
            'Note': 'Saham berisiko tinggi. Tidak direkomendasikan entry.'
        }

    if "Jangka Panjang" in horizon:
        entry_price = min(price, fair_value * 0.95) if fair_value > 0 else price
        tp_price = entry_price * 1.20
        sl_price = entry_price * 0.92
        note = "Strategi Value Investing: Buy on Weakness / Cicil Bertahap."
    else:
        entry_price = price
        tp_price = entry_price * 1.10
        sl_price = entry_price * 0.95
        note = "Strategi Trading Swing / Momentum: Ketat menjaga Stop Loss."

    potential_return = ((tp_price - entry_price) / entry_price) * 100
    potential_risk = ((entry_price - sl_price) / entry_price) * 100
    rrr = potential_return / potential_risk if potential_risk > 0 else 0

    return {
        'Entry': round(entry_price),
        'TP': round(tp_price),
        'SL': round(sl_price),
        'ReturnPct': round(potential_return, 2),
        'RiskPct': round(potential_risk, 2),
        'RRR': f"1 : {rrr:.2f}",
        'Note': note
    }

# ------------------------------------------------------------------------------
# 4. INTERFACE & DISPLAY ENGINE (IPYWIDGETS)
# ------------------------------------------------------------------------------
ticker_widget = widgets.Text(value="BBCA", description="Kode Saham:", placeholder="Contoh: BBCA, TLKM")
btn_web = widgets.Button(description="🌐 Tarik Data dari Web", button_style="info", layout=widgets.Layout(width="220px"))

file_upload = widgets.FileUpload(accept='.pdf', multiple=False, description="📄 Upload PDF IDX")
btn_pdf = widgets.Button(description="📑 Ekstrak dari PDF", button_style="warning", layout=widgets.Layout(width="220px"))

output_area = widgets.Output()

def run_analysis_pipeline(data):
    with output_area:
        clear_output()
        if not data:
            print("❌ Data tidak ditemukan atau gagal diproses.")
            return

        # 1. Scoring Calculation
        scores = {}
        scores["Likuiditas"] = 15 if data['CR'] >= 1.5 else (10 if data['CR'] >= 1.0 else 0)
        scores["Solvabilitas"] = 20 if data['DER'] <= 0.8 else (10 if data['DER'] <= 1.2 else 0)

        roe_s = 15 if data['ROE'] >= 12.0 else (8 if data['ROE'] >= 8.0 else 0)
        npm_s = 10 if data['NPM'] >= 10.0 else (5 if data['NPM'] >= 5.0 else 0)
        scores["Profitabilitas"] = roe_s + npm_s

        scores["Pertumbuhan"] = (data['GrowthYears'] / 4.0) * 20
        scores["Valuasi"] = 20 if data['PER'] < data['PER_Ind'] else 10

        total_score = sum(scores.values())
        is_good = total_score >= 70
        category = "BAIK (FEASIBLE)" if is_good else "KURANG BAIK (RISKY)"

        horizon = "Investasi Jangka Panjang" if (is_good and data['ROE'] >= 12 and data['DER'] <= 1.0 and data['HasDiv']) else "Jangka Pendek / Momentum"

        # 2. Reasoning & Strategy Calculation
        pros, cons = generate_analysis_reasons(data, scores, total_score)
        strat = calculate_trade_strategy(data, is_good, horizon)

        # Build HTML Bullet Lists for Reasons
        pros_html = "".join([f"<li>{p}</li>" for p in pros]) if pros else "<li>Tidak ada keunggulan dominan.</li>"
        cons_html = "".join([f"<li>{c}</li>" for c in cons]) if cons else "<li>Tidak ada catatan risiko kritis.</li>"

        # 3. Render HTML Summary
        badge_col = "#16a34a" if is_good else "#dc2626"
        price_display = f"Rp {data['CurrentPrice']:,.0f}" if data['CurrentPrice'] > 0 else "N/A"

        display(HTML(f"""
        <div style="padding: 15px; border-radius: 8px; background-color: #f8fafc; border: 1px solid #cbd5e1; margin-top: 15px;">
            <h3 style="margin: 0; color: #0f172a;">📊 Hasil Analisis Saham: {data['Ticker']} <small>({data['Source']})</small></h3>
            <p><b>Harga Saat Ini:</b> <span style="font-size: 16px; font-weight: bold;">{price_display}</span></p>
            <hr>
            <p><b>Skor Total Fundamental:</b> <span style="font-size: 18px; font-weight: bold; color: {badge_col};">{total_score:.1f} / 100</span></p>
            <p><b>Kategori Saham:</b> <span style="padding: 3px 8px; border-radius: 4px; color: white; background-color: {badge_col}; font-weight: bold;">{category}</span></p>
            <p><b>Rekomendasi Horizon:</b> <u>{horizon}</u></p>
        </div>

        <!-- CARD RINGKASAN ALASAN & DRIVERS -->
        <div style="padding: 15px; border-radius: 8px; background-color: #ffffff; border: 1px solid #e2e8f0; margin-top: 10px;">
            <h4 style="margin-top: 0; color: #334155;">📌 Ringkasan Alasan Klasifikasi</h4>
            <div style="display: flex; gap: 20px;">
                <div style="flex: 1; background-color: #f0fdf4; padding: 10px; border-radius: 6px; border-left: 4px solid #16a34a;">
                    <h5 style="margin-top: 0; color: #15803d;">✅ Poin Keunggulan (Faktor Positif):</h5>
                    <ul style="margin-bottom: 0; padding-left: 18px; font-size: 13px; color: #1e293b;">
                        {pros_html}
                    </ul>
                </div>
                <div style="flex: 1; background-color: #fef2f2; padding: 10px; border-radius: 6px; border-left: 4px solid #dc2626;">
                    <h5 style="margin-top: 0; color: #b91c1c;">⚠️ Poin Catatan Risiko (Faktor Negatif):</h5>
                    <ul style="margin-bottom: 0; padding-left: 18px; font-size: 13px; color: #1e293b;">
                        {cons_html}
                    </ul>
                </div>
            </div>
        </div>

        <!-- CARD REKOMENDASI TRADING / INVESTASI -->
        <div style="padding: 15px; border-radius: 8px; background-color: #eff6ff; border: 1px solid #93c5fd; margin-top: 10px;">
            <h4 style="margin-top: 0; color: #1e40af;">🎯 Rekomendasi Eksekusi Transaksi</h4>
            <table style="width: 100%; border-collapse: collapse; font-size: 14px;">
                <tr style="border-bottom: 1px solid #cbd5e1;">
                    <td style="padding: 6px;"><b>Harga Masuk (Entry):</b></td>
                    <td style="padding: 6px; color: #1d4ed8; font-weight: bold;">Rp {strat['Entry']:,}</td>
                    <td style="padding: 6px;"><b>Potensi Return:</b></td>
                    <td style="padding: 6px; color: #15803d; font-weight: bold;">+{strat['ReturnPct']}%</td>
                </tr>
                <tr style="border-bottom: 1px solid #cbd5e1;">
                    <td style="padding: 6px;"><b>Target Profit (TP):</b></td>
                    <td style="padding: 6px; color: #15803d; font-weight: bold;">Rp {strat['TP']:,}</td>
                    <td style="padding: 6px;"><b>Potensi Risiko:</b></td>
                    <td style="padding: 6px; color: #b91c1c; font-weight: bold;">-{strat['RiskPct']}%</td>
                </tr>
                <tr>
                    <td style="padding: 6px;"><b>Stop Loss (SL):</b></td>
                    <td style="padding: 6px; color: #b91c1c; font-weight: bold;">Rp {strat['SL']:,}</td>
                    <td style="padding: 6px;"><b>Risk / Reward Ratio:</b></td>
                    <td style="padding: 6px; font-weight: bold;">{strat['RRR']}</td>
                </tr>
            </table>
            <p style="margin-bottom: 0; margin-top: 8px; font-size: 12px; color: #475569;">💡 <i>{strat['Note']}</i></p>
        </div>
        """))

        # Render Visual Bar Chart
        fig, ax = plt.subplots(figsize=(7, 2.5), dpi=100)
        ax.bar(list(scores.keys()), list(scores.values()), color="#2563eb")
        ax.set_ylim(0, 30)
        ax.set_ylabel("Poin Diraih")
        ax.set_title(f"Breakdown Skor Fundamental - {data['Ticker']}")
        plt.tight_layout()
        plt.show()

# Event Handlers
def on_web_click(b):
    data = fetch_data_from_web(ticker_widget.value)
    run_analysis_pipeline(data)

def on_pdf_click(b):
    if file_upload.value:
        uploaded_file = list(file_upload.value.values())[0]
        with open("temp_report.pdf", "wb") as f:
            f.write(uploaded_file['content'])

        data = parse_pdf_financial_report("temp_report.pdf")
        run_analysis_pipeline(data)
    else:
        with output_area:
            clear_output()
            print("⚠️ Silakan pilih/upload file PDF Laporan Keuangan terlebih dahulu!")

btn_web.on_click(on_web_click)
btn_pdf.on_click(on_pdf_click)

# Display UI Layout
layout = widgets.VBox([
    widgets.HTML("<h2>🤖 Analisis Saham + Ringkasan Alasan Kategori</h2>"),
    widgets.HBox([widgets.VBox([widgets.HTML("<b>Opsi 1: Ambil dari Web</b>"), ticker_widget, btn_web]),
                  widgets.VBox([widgets.HTML("<b>Opsi 2: Upload PDF IDX</b>"), file_upload, btn_pdf])]),
    output_area
])

display(layout)
```
